## Mục tiêu
Làm sạch dữ liệu merged_data.csv bằng SQLite và Python:
- Đưa dữ liệu vào SQLite và kiểm tra dữ liệu.
- Kiểm tra số lượng và tỷ lệ dữ liệu thiếu trước khi cleaning.
- Xử lý dữ liệu thiếu phù hợp với từng loại biến.
- Kiểm tra dữ liệu trùng lặp và các giá trị bất hợp lệ.
- So sánh dữ liệu trước và sau khi cleaning.
- Xuất dữ liệu sau khi làm sạch thành cleaned_data.csv để thực hiện EDA.

## Luồng dữ liệu

`merged_data.csv` → **SQLite** → đọc dữ liệu từ SQLite → **Missing Assessment** → **Imputation** → **Validation** → `cleaned_data.csv`

> `cleaned_data.csv` là đầu ra sạch để Hải và Tài tiếp tục thực hiện Univariate EDA và Multivariate EDA.


## 1. Import thư viện

```python
import pandas as pd
import numpy as np
import sqlite3
import os
import matplotlib.pyplot as plt
```

## 2. Đọc dữ liệu đầu vào từ Yến/Linh

Nguồn đầu vào chính của phần này là `merged_data.csv` – dữ liệu đã được Yến tích hợp từ Unified Static Master và Dynamic Data.


```python
INPUT_FILE = "merged_data.csv"

df = pd.read_csv(INPUT_FILE)

print("Kích thước dữ liệu:", df.shape)
print("Số dòng:", len(df))
print("Số cột:", len(df.columns))

display(df.head())
```

## 3. Kiểm tra dữ liệu trước khi đưa vào SQLite

```python
print("Tên các cột:")
print(df.columns.tolist())

print("\nKiểu dữ liệu:")
display(df.dtypes.to_frame("dtype"))

print("\nSố dòng trùng:", df.duplicated().sum())

print("\nGiá trị thiếu:")
missing_before = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percent": (df.isna().mean() * 100).round(2)
}).sort_values("Missing_Count", ascending=False)

display(missing_before)
```

## 4. Database 2 – SQLite

Sử dụng SQLite làm cơ sở dữ liệu quan hệ thứ hai của nhóm.

- Database: `tv_products.db`
- Table: `merged_data`
- Dữ liệu được nạp trực tiếp từ `merged_data.csv`.


```python
DB_FILE = "tv_products.db"
TABLE_NAME = "merged_data"

conn = sqlite3.connect(DB_FILE)

# Ghi dữ liệu vào SQLite
df.to_sql(
    TABLE_NAME,
    conn,
    if_exists="replace",
    index=False
)

print("Đã tạo/nạp bảng SQLite:", TABLE_NAME)
```

## 5. Kiểm tra dữ liệu trong SQLite

```python
# Danh sách bảng
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

display(tables)

# Đếm số records
record_count = pd.read_sql_query(
    f"SELECT COUNT(*) AS total_records FROM {TABLE_NAME};",
    conn
)

display(record_count)

# Xem 5 records đầu
sample_sql = pd.read_sql_query(
    f"SELECT * FROM {TABLE_NAME} LIMIT 5;",
    conn
)

display(sample_sql)
```

## 6. Đọc dữ liệu trở lại từ SQLite

Bước này chứng minh dữ liệu sau khi lưu vào SQLite có thể được kết nối và đọc lại bằng Python.


```python
df_sql = pd.read_sql_query(
    f"SELECT * FROM {TABLE_NAME}",
    conn
)

print("Kích thước đọc từ SQLite:", df_sql.shape)
display(df_sql.head())
```

## 7. Đóng kết nối SQLite sau khi đọc dữ liệu

```python
conn.close()

print("Đã đóng kết nối SQLite.")
```

# PHẦN CLEANING

## 8. Dataset trước Imputation

Tạo một bản sao dữ liệu vừa đọc từ SQLite để giữ nguyên dữ liệu trước xử lý, phục vụ so sánh Before/After.


```python
df_before = df_sql.copy()
df_clean = df_sql.copy()

print("Dataset trước Cleaning:", df_before.shape)
display(df_before.head())
```

## 9. Thống kê Missing: số lượng và tỷ lệ %

Theo yêu cầu Cleaning, kiểm tra số giá trị thiếu trên từng cột trước khi Imputation.


```python
missing_before = pd.DataFrame({
    "Missing_Count": df_before.isna().sum(),
    "Missing_Percent": (df_before.isna().mean() * 100).round(2)
}).sort_values("Missing_Count", ascending=False)

display(missing_before)
```

## 1. Chuẩn hóa kiểu dữ liệu trước Imputation

Chuẩn hóa các cột có ý nghĩa số/ngày trước khi phân loại Numeric và Categorical.

- `release_year`: chuyển sang numeric.
- `data_date`: chuyển sang datetime.


```python
numeric_cols = df_clean.select_dtypes(include=np.number).columns.tolist()

categorical_cols = df_clean.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numeric columns:")
print(numeric_cols)

print("\nCategorical columns:")
print(categorical_cols)
```

## 2. Phân loại cột sau khi chuẩn hóa

### Nguyên tắc
- Numeric → median.
- Categorical → `Unknown`.
- `data_date` đã là datetime và sẽ được kiểm tra riêng.


```python
# Chuẩn hóa kiểu dữ liệu
df_clean["release_year"] = pd.to_numeric(
    df_clean["release_year"],
    errors="coerce"
)

df_clean["data_date"] = pd.to_datetime(
    df_clean["data_date"],
    errors="coerce"
)

numeric_cols = df_clean.select_dtypes(include=np.number).columns.tolist()

categorical_cols = df_clean.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numeric columns:")
print(numeric_cols)

print("\nCategorical columns:")
print(categorical_cols)
```

## 3. Imputation cho Numeric

Dùng **median** cho các cột số có giá trị thiếu.

Cách này phù hợp hơn mean trong trường hợp dữ liệu sản phẩm có thể chứa giá trị lớn bất thường.


```python
numeric_imputation = {}

for col in numeric_cols:
    missing_count = df_clean[col].isna().sum()

    if missing_count > 0:
        median_value = df_clean[col].median()
        numeric_imputation[col] = median_value
        df_clean[col] = df_clean[col].fillna(median_value)

print("Các cột numeric đã Imputation:")
for col, value in numeric_imputation.items():
    print(f"- {col}: median = {value}")
```

## 4. Imputation cho Categorical

Dùng giá trị **`Unknown`** cho dữ liệu phân loại bị thiếu.

Lý do: nhiều trường của Walmart không tương ứng trực tiếp với các thuộc tính được crawl từ Điện Máy Xanh. Vì vậy, gán mode có thể tạo ra giá trị giả và làm sai phân bố nhóm.


```python
categorical_imputation = []

for col in categorical_cols:
    missing_count = df_clean[col].isna().sum()

    if missing_count > 0:
        df_clean[col] = df_clean[col].fillna("Unknown")
        categorical_imputation.append(col)

print("Các cột categorical đã thay missing bằng 'Unknown':")
print(categorical_imputation)
```

## 5. Kiểm tra kiểu dữ liệu sau chuẩn hóa

```python
print("Kiểu dữ liệu sau chuẩn hóa:")
display(df_clean.dtypes.to_frame("dtype"))
```

## 6. Validation – kiểm tra sau Imputation

Mục tiêu: xác nhận Dataset đã đi qua bước **Missing → Imputation → Validation**.


```python
missing_after = pd.DataFrame({
    "Missing_Count": df_clean.isna().sum(),
    "Missing_Percent": (df_clean.isna().mean() * 100).round(2)
}).sort_values("Missing_Count", ascending=False)

display(missing_after)

print("Tổng số missing trước:", int(df_before.isna().sum().sum()))
print("Tổng số missing sau:", int(df_clean.isna().sum().sum()))
print("Số dòng trùng sau Cleaning:", int(df_clean.duplicated().sum()))
```

## 7. Bảng Before / After

So sánh số lượng giá trị thiếu trước và sau Imputation.


```python
validation_report = pd.DataFrame({
    "Before_Missing": df_before.isna().sum(),
    "After_Missing": df_clean.isna().sum()
})

validation_report["Filled_Values"] = (
    validation_report["Before_Missing"]
    - validation_report["After_Missing"]
)

validation_report["Before_Missing_Percent"] = (
    df_before.isna().mean() * 100
).round(2)

validation_report["After_Missing_Percent"] = (
    df_clean.isna().mean() * 100
).round(2)

display(validation_report)
```

## 8. So sánh thống kê Numeric trước và sau Imputation

So sánh Mean, Median và Standard Deviation để kiểm tra mức thay đổi của phân bố sau khi điền missing.


```python
numeric_compare_cols = [
    col for col in numeric_cols
    if col in df_before.columns
]

before_stats = df_before[numeric_compare_cols].describe().T[
    ["mean", "50%", "std"]
].rename(columns={
    "mean": "Before_Mean",
    "50%": "Before_Median",
    "std": "Before_Std"
})

after_stats = df_clean[numeric_compare_cols].describe().T[
    ["mean", "50%", "std"]
].rename(columns={
    "mean": "After_Mean",
    "50%": "After_Median",
    "std": "After_Std"
})

stats_comparison = before_stats.join(after_stats)

display(stats_comparison)
```

## 9. So sánh phân bố trước và sau Imputation – Price

Biểu đồ dùng để kiểm tra trực quan sự thay đổi của phân bố `price` sau khi điền giá trị thiếu bằng median.


```python
plt.figure(figsize=(10, 5))

plt.hist(
    df_before["price"].dropna(),
    bins=30,
    alpha=0.6,
    label="Before Imputation"
)

plt.hist(
    df_clean["price"],
    bins=30,
    alpha=0.6,
    label="After Imputation"
)

plt.title("Price Distribution Before vs After Imputation")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

## 10. Validation dữ liệu đầu ra

Kiểm tra các điều kiện cơ bản trước khi xuất Master Dataset:

1. Không còn missing.
2. Không có dòng trùng.
3. Các trường số quan trọng không âm.
4. Rating nằm trong khoảng hợp lệ 0–5.


```python
# Kiểm tra missing
total_missing = int(df_clean.isna().sum().sum())

# Kiểm tra duplicate
total_duplicates = int(df_clean.duplicated().sum())

# Kiểm tra giá trị âm
negative_price = int((df_clean["price"] < 0).sum())
negative_views = int((df_clean["monthly_views"] < 0).sum())
negative_orders = int((df_clean["monthly_orders"] < 0).sum())
negative_stock = int((df_clean["stock_quantity"] < 0).sum())

# Rating hợp lệ
invalid_customer_rating = int(
    (~df_clean["customer_rating"].between(0, 5)).sum()
)

validation_summary = pd.DataFrame({
    "Check": [
        "Total missing",
        "Duplicate rows",
        "Negative price",
        "Negative monthly_views",
        "Negative monthly_orders",
        "Negative stock_quantity",
        "Invalid customer_rating"
    ],
    "Count": [
        total_missing,
        total_duplicates,
        negative_price,
        negative_views,
        negative_orders,
        negative_stock,
        invalid_customer_rating
    ]
})

display(validation_summary)
```

## 11. Master Dataset hoàn chỉnh

Đây là dataset sau khi hoàn thành Cleaning và Validation.


```python
master_dataset = df_clean.copy()

print("Kích thước Master Dataset:", master_dataset.shape)
print("Tổng missing:", master_dataset.isna().sum().sum())
print("Tổng duplicate:", master_dataset.duplicated().sum())

display(master_dataset.head())
```

## 12. Xuất `cleaned_data.csv`

Đây là **đầu ra chính của phần Dương**, được bàn giao cho:

- **Bạn Hải:** Univariate EDA
- **Bạn Tài:** Multivariate EDA

Hai bạn sử dụng `cleaned_data.csv` làm dataset đầu vào để toàn bộ nhóm phân tích trên cùng một phiên bản dữ liệu sạch.


```python
OUTPUT_FILE = "cleaned_data.csv"

master_dataset.to_csv(
    OUTPUT_FILE,
    index=False,
    encoding="utf-8-sig"
)

print(f"Đã xuất: {OUTPUT_FILE}")
```

## 13. Đọc lại file sau khi xuất – Final Validation

```python
final_df = pd.read_csv(OUTPUT_FILE)

print("FINAL VALIDATION")
print("-" * 40)
print("Số dòng:", final_df.shape[0])
print("Số cột:", final_df.shape[1])
print("Tổng missing:", int(final_df.isna().sum().sum()))
print("Số dòng trùng:", int(final_df.duplicated().sum()))

display(final_df.head())
```

# 04_Hai – UNIVARIATE EDA

**Thành viên:** Hải  
**Nối tiếp:** Yến – Prepare Data + Data Integration → Linh – Qdrant → Dương – SQLite + Cleaning → Hải – Univariate EDA

## Mục tiêu
Phân tích dữ liệu `cleaned_data.csv` theo yêu cầu Univariate EDA:
- Numerical Variables: Mean, Median, Standard Deviation, Skewness, Kurtosis, Distribution và Outlier bằng IQR.
- Categorical Variables: Frequency và Percentage.
- Biểu đồ: Histogram + KDE, Boxplot, Bar chart và Pie chart khi biến có dưới 5 nhóm.
- Chọn các biến đại diện để tránh biểu đồ quá dàn trải.
- Mỗi biểu đồ có Title, X, Y, Legend và phần **So What?**.

## 1. Import thư viện và đọc dữ liệu

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.patches import Patch

df = pd.read_csv("cleaned_data.csv")

print("Kích thước dữ liệu:", df.shape)
display(df.head())
```

## 2. Xác định biến phân tích

Để tập trung phân tích chuyên sâu và tránh biểu đồ quá dàn trải, nhóm lựa chọn 3 biến số đại diện:
- `price`
- `monthly_views`
- `monthly_orders`

Các biến phân loại được chọn:
- `manufacturer`
- `operating_system`
- `source`


```python
numeric_cols = ["price", "monthly_views", "monthly_orders"]
categorical_cols = ["manufacturer", "operating_system", "source"]

print("Numerical Variables:", numeric_cols)
print("Categorical Variables:", categorical_cols)
```

# 3. Numerical Variables

## 3.1. Mean, Median, Standard Deviation, Skewness và Kurtosis

```python
numeric_stats = pd.DataFrame({
    "Mean": df[numeric_cols].mean(),
    "Median": df[numeric_cols].median(),
    "Standard Deviation": df[numeric_cols].std(),
    "Skewness": df[numeric_cols].skew(),
    "Kurtosis": df[numeric_cols].kurt()
}).round(2)

display(numeric_stats)
```

**So What?**

Bảng thống kê cho biết mức trung tâm, mức độ phân tán và hình dạng phân phối của 3 biến số. Các chỉ số này là cơ sở để lựa chọn cách trực quan hóa và xử lý dữ liệu phù hợp ở các bước phân tích tiếp theo.

## 3.2. Distribution – Histogram + KDE

Histogram kết hợp KDE được dùng để quan sát hình dạng phân phối của từng biến số.

```python
for col in numeric_cols:
    plt.figure(figsize=(8, 5))
    sns.histplot(data=df, x=col, kde=True, label=col)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.legend(title="Variable")
    plt.tight_layout()
    plt.show()

    print(f"**So What? – {col}:** Biểu đồ cho thấy hình dạng phân phối của {col}, mức độ tập trung và độ lệch của dữ liệu. Có thể đối chiếu với Skewness và Kurtosis trong bảng thống kê để nhận diện phân phối bất đối xứng hoặc có đuôi dài.")
```

## 3.3. Outlier – IQR

Outlier được phát hiện bằng phương pháp IQR theo công thức:
- IQR = Q3 − Q1
- Lower Bound = Q1 − 1.5 × IQR
- Upper Bound = Q3 + 1.5 × IQR

```python
outlier_rows = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_rows.append({
        "Variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower,
        "Upper Bound": upper,
        "Outlier Count": int(outliers)
    })

outlier_table = pd.DataFrame(outlier_rows).round(2)
display(outlier_table)
```

## 3.4. Boxplot

Boxplot được sử dụng để trực quan hóa phân bố và kiểm tra các giá trị ngoại lệ theo IQR.

```python
for col in numeric_cols:
    plt.figure(figsize=(8, 5))

    sns.boxplot(y=df[col])

    plt.title(f"Boxplot of {col}")
    plt.xlabel("Variable")
    plt.ylabel(col)

    # Legend để đáp ứng yêu cầu trình bày biểu đồ
    legend_handle = Patch(label=col)
    plt.legend(handles=[legend_handle], title="Variable")

    plt.tight_layout()
    plt.show()

    print(f"**So What? – {col}:** Boxplot cho thấy trung vị, khoảng tứ phân vị và các giá trị nằm ngoài whisker. Những điểm nằm ngoài whisker là các quan sát có khả năng là outlier và cần được lưu ý khi phân tích.")
```

# 4. Categorical Variables

## 4.1. Frequency và Percentage

Frequency và Percentage được tính cho các biến phân loại đã chọn. Với các biến có nhiều nhóm, biểu đồ chỉ hiển thị Top 10 để dễ quan sát.

```python
for col in categorical_cols:
    frequency = df[col].value_counts(dropna=False)
    percentage = (df[col].value_counts(normalize=True, dropna=False) * 100).round(2)

    summary = pd.DataFrame({
        "Frequency": frequency,
        "Percentage": percentage
    })

    print(f"===== {col} =====")
    display(summary.head(10))
```

## 4.2. Bar chart – Frequency

```python
for col in categorical_cols:
    data = df[col].value_counts(dropna=False).head(10)

    plt.figure(figsize=(9, 5))
    data.plot(kind="bar", label="Frequency")

    plt.title(f"Frequency of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.xticks(rotation=45, ha="right")
    plt.legend(title="Measure")
    plt.tight_layout()
    plt.show()

    print(f"**So What? – {col}:** Bar chart cho thấy các nhóm xuất hiện với tần suất khác nhau như thế nào và nhóm nào chiếm tỷ trọng cao trong dữ liệu.")
```

## 4.3. Percentage – Bar chart

```python
for col in categorical_cols:
    data = (df[col].value_counts(normalize=True, dropna=False) * 100).round(2).head(10)

    plt.figure(figsize=(9, 5))
    data.plot(kind="bar", label="Percentage")

    plt.title(f"Percentage of {col}")
    plt.xlabel(col)
    plt.ylabel("Percentage (%)")
    plt.xticks(rotation=45, ha="right")
    plt.legend(title="Measure")
    plt.tight_layout()
    plt.show()

    print(f"**So What? – {col}:** Biểu đồ cho biết tỷ lệ của từng nhóm trong dữ liệu, giúp nhận diện nhóm chiếm tỷ trọng lớn hoặc các nhóm có tỷ lệ thấp.")
```

## 4.4. Pie chart

Theo yêu cầu, Pie chart chỉ sử dụng khi biến có dưới 5 nhóm. `source` có 2 nhóm nên phù hợp để biểu diễn tỷ lệ.

```python
source_percentage = df["source"].value_counts(normalize=True) * 100

plt.figure(figsize=(6, 6))
plt.pie(
    source_percentage,
    labels=source_percentage.index,
    autopct="%1.1f%%"
)

plt.title("Percentage of source")
plt.legend(title="Source")
plt.tight_layout()
plt.show()

print("**So What – source:** Biểu đồ cho thấy tỷ trọng của hai nguồn dữ liệu trong tập `cleaned_data.csv`, giúp kiểm tra mức độ cân bằng giữa các nguồn.")
```

# 5. Tổng kết Univariate EDA

- **Numerical:** đã tính Mean, Median, Standard Deviation, Skewness và Kurtosis; phân tích Distribution bằng Histogram + KDE; phát hiện Outlier bằng IQR và trực quan hóa bằng Boxplot.
- **Categorical:** đã tính Frequency và Percentage; trực quan hóa bằng Bar chart; dùng Pie chart cho `source` vì biến này có dưới 5 nhóm.
- Các biểu đồ được chọn từ các biến đại diện để tập trung vào nội dung chính thay vì vẽ toàn bộ cột trong dataset.
