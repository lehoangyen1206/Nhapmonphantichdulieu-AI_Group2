```python
import pandas as pd
import numpy as np
```

## 2. Đọc dữ liệu Static

```python
dmx = pd.read_csv("tivi.csv")
wm = pd.read_csv("walmart_tv.csv")

print("Điện Máy Xanh:", dmx.shape)
print("Walmart:", wm.shape)

display(dmx.head())
display(wm.head())
```

## 3. Kiểm tra dữ liệu

```python
print("Cột Điện Máy Xanh:")
print(dmx.columns.tolist())

print("\nCột Walmart:")
print(wm.columns.tolist())

print("\nSố dòng trùng:")
print("Điện Máy Xanh:", dmx.duplicated().sum())
print("Walmart:", wm.duplicated().sum())
```

## 4. Đồng bộ tên cột

```python
dmx = dmx.rename(columns={
    "product_name": "product_name",
    "sale_price": "price",
    "tv_type": "tv_type",
    "screen_size": "screen_size",
    "resolution": "resolution",
    "screen_type": "screen_type",
    "operating_system": "operating_system",
    "ram": "ram",
    "rom": "rom",
    "release_year": "release_year"
})

wm = wm.rename(columns={
    "Title": "product_name",
    "Price": "price",
    "Manufacturer": "manufacturer",
    "Model Name": "model_name",
    "Screen Size": "screen_size",
    "Average Rating": "rating",
    "Num Of Reviews": "num_reviews",
    "Specifications": "specifications"
})

print("Đã chuẩn hóa tên cột.")
```

## 5. pd.concat() – Unified Static Master Table

```python
cols = [
    "product_name", "price", "tv_type", "screen_size",
    "resolution", "screen_type", "operating_system",
    "ram", "rom", "release_year",
    "manufacturer", "model_name", "rating",
    "num_reviews", "specifications", "source"
]

dmx = dmx.reindex(columns=cols[:-1])
dmx["source"] = "dienmayxanh"

wm = wm.reindex(columns=cols[:-1])
wm["source"] = "walmart"

unified_static_master = pd.concat(
    [dmx, wm],
    ignore_index=True
)

print("Kích thước:", unified_static_master.shape)
display(unified_static_master.head())
```

## 6. Tạo Dynamic Data

```python
np.random.seed(42)

dynamic = pd.DataFrame()
dynamic["product_name"] = unified_static_master["product_name"]
dynamic["monthly_views"] = np.random.randint(100, 10000, len(unified_static_master))
dynamic["monthly_orders"] = np.random.randint(5, 500, len(unified_static_master))
dynamic["customer_rating"] = np.round(
    np.random.uniform(3.0, 5.0, len(unified_static_master)), 1
)
dynamic["stock_quantity"] = np.random.randint(0, 200, len(unified_static_master))
dynamic["data_date"] = pd.date_range(
    start="2026-01-01",
    periods=len(unified_static_master),
    freq="D"
)

display(dynamic.head())
print("Kích thước:", dynamic.shape)
```

```python
dynamic.to_excel(
    "F:/AI code/dynamic_data.xlsx",
    index=False
)
print("Đã xuất dynamic_data.xlsx")
```

## 7. pd.merge() – Tích hợp Static và Dynamic Data

**Loại JOIN: Left Join**

Chọn Left Join để giữ toàn bộ sản phẩm trong Unified Static Master Table. Nếu thiếu dữ liệu động tương ứng thì sản phẩm tĩnh vẫn được giữ lại.

```python
merged_data = pd.merge(
    unified_static_master,
    dynamic,
    on="product_name",
    how="left"
)

print("Kích thước sau merge:", merged_data.shape)
display(merged_data.head())
```

## Mục tiêu
Xử lý dữ liệu merged_data.csv trên cơ sở dữ liệu Qdrant:
- Đưa dữ liệu vào Qdrant và tổ chức dữ liệu để lưu trữ.
- Thực hiện kiểm tra, truy vấn và tìm kiếm dữ liệu trên Qdrant.
- Kết nối Qdrant với Python để thao tác và khai thác dữ liệu.
- Kiểm tra kết quả lưu trữ và truy vấn dữ liệu.

```python
import pandas as pd

df = pd.read_csv("merged_data.csv")

print("Số records:", len(df))
print("Số columns:", len(df.columns))

display(df.head())
```

```python
#chọn dữ liệu để embedding
text_columns = [
    "product_name",
    "tv_type",
    "screen_size",
    "resolution",
    "screen_type",
    "operating_system",
    "ram",
    "rom",
    "release_year"
]
```

```python
def create_text(row):
    return (
        f"Sản phẩm: {row['product_name']}. "
        f"Loại TV: {row['tv_type']}. "
        f"Kích thước: {row['screen_size']}. "
        f"Độ phân giải: {row['resolution']}. "
        f"Loại màn hình: {row['screen_type']}. "
        f"Hệ điều hành: {row['operating_system']}. "
        f"RAM: {row['ram']}. "
        f"ROM: {row['rom']}. "
        f"Năm ra mắt: {row['release_year']}."
    )

df["text"] = df.apply(create_text, axis=1)
display(df[["product_name", "text"]].head())
```

```python
#tạo embedding
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(
    df["text"].tolist(),
    show_progress_bar=True
)

print("Số vector:", len(embeddings))
print("Số chiều:", len(embeddings[0]))
```

```python
#kết nối qdrant
from qdrant_client import QdrantClient

client = QdrantClient(path="./qdrant_storage_linh")

collection_name = "tv_products"
```

```python
# Tạo Collection
from qdrant_client.models import Distance, VectorParams

if client.collection_exists(collection_name):
    client.delete_collection(collection_name)

client.create_collection(
    collection_name=collection_name,
    vectors_config=VectorParams(
        size=384,
        distance=Distance.COSINE
    )
)
```

```python
# Tạo Point
from qdrant_client.models import PointStruct

payloads = (
    df.drop(columns=["text"])
      .fillna("")
      .to_dict(orient="records")
)

points = []

for i, (embedding, payload) in enumerate(
    zip(embeddings, payloads)
):
    points.append(
        PointStruct(
            id=i,
            vector=embedding.tolist(),
            payload=payload
        )
    )
```

```python
# Ingest vào Qdrant
client.upsert(
    collection_name=collection_name,
    points=points
)

print("Đã ingest dữ liệu vào Qdrant!")
```

```python
# Kiểm tra số lượng records
info = client.get_collection(collection_name)

print("Số records trong Qdrant:",
      info.points_count)
```

```python
# Lấy record mẫu
records, next_page = client.scroll(
    collection_name=collection_name,
    limit=5,
    with_payload=True,
    with_vectors=False
)

for record in records:
    print("ID:", record.id)
    print("Product:", record.payload.get("product_name"))
    print("Price:", record.payload.get("sale_price"))
    print("-" * 50)
```

```python
# Query / Similarity Search
query = "Tivi LG 4K 65 inch"

query_vector = model.encode(query).tolist()

results = client.query_points(
    collection_name=collection_name,
    query=query_vector,
    limit=5,
    with_payload=True
).points
```

```python
for result in results:
    print("Score:", round(result.score, 4))
    print("Product:",
          result.payload.get("product_name"))
    print("Price:",
          result.payload.get("sale_price"))
    print("Resolution:",
          result.payload.get("resolution"))
    print("-" * 50)
```
